**Preprocessing of Dataset1 (Brench/Numina-Math-R1-COT-172K)**

```python
from datasets import load_dataset
import re

# Load the Numina‑Math CoT dataset (R1 release)
dataset1 = load_dataset("Brench/Numina-Math-R1-COT-172K")
all_columns = dataset1["train"].column_names

# ChatML template kept as a single triple‑quoted string
new_template = """<|im_start|>system\n{system_prompt}<|im_end|>\n<|im_start|>user\n{user_prompt}<|im_end|>\n<|im_start|>assistant\n{model_answer}<|im_end|>\n"""


# Keep only the problem statement and the re‑annotated assistant answer
keep_cols = ["problem", "reannotated_assistant_content"]
remove_cols = [col for col in all_columns if col not in keep_cols]

dataset1 = dataset1.remove_columns(remove_cols)

# Filter out overly long answers (>2000 chars) and questions (>900 chars)
dataset1 = dataset1.filter(
    lambda x: len(x["reannotated_assistant_content"]) < 2000,
    num_proc=4,
)

dataset1 = dataset1.filter(
    lambda x: len(x["problem"]) < 900,
    num_proc=4,
)

# Regex helpers for cleaning the assistant content
box_pat = re.compile(
    r"\*\*Final Answer:\*\*.*?\\boxed\{\\?\$?([^}]+)\}.*?",
    flags=re.DOTALL,
)
trail_pat = re.compile(
    r"[ \t]*\\+\]?\s*(?:\n|$)",
    flags=re.MULTILINE,
)

def clean(example):
    """Remove LaTeX cruft and unify the final answer marker."""
    txt = example["reannotated_assistant_content"]
    txt = box_pat.sub(lambda m: f"The Answer = {m.group(1).strip()}", txt)
    txt = trail_pat.sub("\n", txt)
    example["reannotated_assistant_content"] = txt.rstrip()
    return example

dataset1 = dataset1.map(clean, num_proc=4)

# Format each record into ChatML
system_prompt = """You are a highly skilled math problem solver.
Carefully reason through each question step by step in your mind, think smartly about how you can reach the solution with given details, consider approaches
detailed reasoning is **vital** for how you arrive at your conclusions. Only after the completing your reasoning, provide a clear, brief, step by step solution and the final Answer.
"""

def format_dataset(example):
    return {
        "text": new_template.format(
            system_prompt=system_prompt,
            user_prompt=example["problem"],
            model_answer=example["reannotated_assistant_content"],
        )
    }

dataset1 = dataset1.map(format_dataset)

# Retain only the final "text" column
dataset1 = dataset1["train"].remove_columns(
    [c for c in dataset1["train"].column_names if c != "text"]
)

dataset1
```

**Preprocessing of Dataset2 (SkunkworksAI/reasoning-0.01)**

```python
from datasets import load_dataset

# Load the SkunkworksAI reasoning-0.01 dataset
# Splits: train / validation / test
dataset2 = load_dataset("SkunkworksAI/reasoning-0.01")
all_columns = dataset2["train"].column_names

# Keep only the instruction, chain-of-thought (reasoning) and final output
keep_cols = ["instruction", "reasoning", "output"]
remove_cols = [c for c in all_columns if c not in keep_cols]

dataset2 = dataset2.remove_columns(remove_cols)

# Filter overly long fields to stay within context window
# -------------------------------------------------------
#   reasoning  < 2000 chars
#   output     < 1000 chars
#   instruction<  900 chars

dataset2 = dataset2.filter(lambda x: len(x["reasoning"])     < 2000, num_proc=4)
dataset2 = dataset2.filter(lambda x: len(x["output"])        < 1000, num_proc=4)
dataset2 = dataset2.filter(lambda x: len(x["instruction"])   <  900, num_proc=4)

# Wrap each reasoning block with <think> … </think> tags so downstream
# reward functions can easily locate the chain-of-thought text.
# We modify batches for speed (batched=True).

def add_tags_batched(examples):
    examples["reasoning"] = [f"<think> {txt} </think>" for txt in examples["reasoning"]]
    return examples

dataset2 = dataset2.map(add_tags_batched, batched=True)

# Build ChatML prompt for every record using the existing `new_template`
# (defined earlier in the notebook / script).

system_prompt = """You are a highly skilled math problem solver.
Carefully reason through each question step by step in your mind, think smartly about how you can reach the solution with given details, consider approaches, detailed reasoning is **vital** for how you arrive at your conclusions. Only after the completing your reasoning, provide a clear, brief, step by step solution and the final Answer.
"""
def format_dataset(example):
    user_prompt  = example["instruction"]
    model_answer = example["reasoning"] + " " + example["output"]

    return {
        "text": new_template.format(
            system_prompt=system_prompt,
            user_prompt=user_prompt,
            model_answer=model_answer,
        )
    }

dataset2 = dataset2.map(format_dataset)

# Keep only the final "text" column (drop helpers)
dataset2 = dataset2["train"].remove_columns(
    [c for c in dataset2["train"].column_names if c != "text"]
)

dataset2
```

**Preprocessing of Dataset3 (Magpie-Align/Magpie-Reasoning-V1-150K-CoT-Deepseek-R1-Llama-70B)**

```python
from datasets import load_dataset
import re

# Load Magpie-Reasoning CoT dataset
dataset3 = load_dataset("Magpie-Align/Magpie-Reasoning-V1-150K-CoT-Deepseek-R1-Llama-70B")

# Keep only the instruction (question) and response (reasoning + answer)
keep_cols = ["instruction", "response"]
all_columns = dataset3["train"].column_names
remove_cols = [c for c in all_columns if c not in keep_cols]
dataset3 = dataset3.remove_columns(remove_cols)

# Trim very long examples to stay within context window
# – assistant response < 2500 chars
# – user instruction < 900 chars
dataset3 = dataset3.filter(lambda x: len(x["response"])    < 2500, num_proc=4)
dataset3 = dataset3.filter(lambda x: len(x["instruction"]) < 900,  num_proc=4)

# Regex: replace "**Final Answer:** … \boxed{…}" with plain text marker
box_pat = re.compile(
    r"\*\*Final Answer(?::)?\*\*"    # "**Final Answer**" (colon optional)
    r".*?"                               # any text until boxed answer
    r"\\boxed\{([^}]+)\}"              # capture payload inside \boxed{}
    r".*?\\\]\s*",                      # consume closing "\]" if present
    flags=re.DOTALL,
)

def to_plain(match: re.Match) -> str:
    raw   = match.group(1)                  # e.g. "$3,\!140" or "990x"
    clean = re.sub(r"[\\!]", "", raw)   # strip backslashes and "!"; keep "$" & digits
    return f"The Answer = {clean.strip()}"

# Apply the substitution across the training split
def fix(example):
    txt = box_pat.sub(to_plain, example["response"])
    example["response"] = txt.rstrip()      # drop trailing newline, if any
    return example

dataset3["train"] = dataset3["train"].map(fix, num_proc=4, desc="Normalize Final Answer blocks")

# Wrap each sample into ChatML format
new_template = """<|im_start|>system\n{system_prompt}<|im_end|>\n<|im_start|>user\n{user_prompt}<|im_end|>\n<|im_start|>assistant\n{model_answer}<|im_end|>\n"""

system_prompt = """You are a highly skilled math problem solver.
Carefully reason through each question step by step in your mind, think smartly about how you can reach the solution with given details, consider approaches
detailed reasoning is **vital** for how you arrive at your conclusions. Only after the completing your reasoning, provide a clear, brief, step by step solution and the final Answer.
"""
def format_dataset(example):
    user_prompt   = example["instruction"]
    model_answer  = example["response"]
    text = new_template.format(
        system_prompt=system_prompt,
        user_prompt=user_prompt,
        model_answer=model_answer,
    )
    return {"text": text}

# Convert every row and drop helper columns
dataset3 = dataset3.map(format_dataset)
dataset3 = dataset3["train"].remove_columns([c for c in dataset3["train"].column_names if c != "text"])
dataset3
```

**Preprocessing of Dataset4 (LangAGI-Lab/magpie-reasoning-v1-20k-thought-summary-math-verifiable)**

```python
from datasets import load_dataset
import re

# Load LangAGI “thought-summary” reasoning dataset (math-verifiable subset)
dataset4 = load_dataset("LangAGI-Lab/magpie-reasoning-v1-20k-thought-summary-math-verifiable")

# Keep only instruction (question), API reasoning, and final response
keep_cols = ["instruction", "o1-mini_api_response", "response"]
all_columns = dataset4["train"].column_names
remove_cols = [c for c in all_columns if c not in keep_cols]

dataset4 = dataset4.remove_columns(remove_cols)

# Drop rows where the intermediate reasoning is missing (None)

def _not_none(example):
    return example["o1-mini_api_response"] is not None

dataset4 = dataset4["train"].filter(_not_none)

# Length filters to fit model context window

dataset4 = dataset4.filter(lambda ex: len(ex["o1-mini_api_response"]) <= 2000)
dataset4 = dataset4.filter(lambda ex: len(ex["instruction"])        <= 900)
dataset4 = dataset4.filter(lambda ex: len(ex["response"])           <= 1000)

# ---------------------------------------------------------------------------
# Normalise answer phrases → replace many variants with “The Answer = …”
# ---------------------------------------------------------------------------

def fix_answer_phrases(example):
    text = example["response"]
    text = re.sub(
        r"The\s+(?:final\s+)?answer\s+is[:]?:?\s*",  # case-insensitive pattern
        "The Answer = ",
        text,
        flags=re.IGNORECASE,
    )
    example["response"] = text
    return example

dataset4 = dataset4.map(fix_answer_phrases)

# Wrap API reasoning with <think> … </think> so reward funcs can detect it

def add_tags_batched(examples):
    examples["o1-mini_api_response"] = [f"<think> {txt} </think>" for txt in examples["o1-mini_api_response"]]
    return examples

dataset4 = dataset4.map(add_tags_batched, batched=True)

# ---------------------------------------------------------------------------
# Format into ChatML using the shared `new_template` defined earlier
# ---------------------------------------------------------------------------

system_prompt = """You are a highly skilled math problem solver.
Carefully reason through each question step by step in your mind, think smartly about how you can reach the solution with given details, consider approaches
detailed reasoning is **vital** for how you arrive at your conclusions. Only after the completing your reasoning, provide a clear, brief, step by step solution and the final Answer.
"""
def format_dataset(example):
    user_prompt  = example["instruction"]
    model_answer = example["o1-mini_api_response"] + " " + example["response"]
    text = new_template.format(
        system_prompt=system_prompt,
        user_prompt=user_prompt,
        model_answer=model_answer,
    )
    return {"text": text}

dataset4 = dataset4.map(format_dataset)

# Keep only the final text column for downstream tokenisation / training
dataset4 = dataset4.remove_columns([c for c in dataset4.column_names if c != "text"])
dataset4
```

**SFT Training**

```python
from transformers import TrainingArguments
from trl import SFTTrainer

# TrainingArguments control all standard HF Trainer flags
training_args = TrainingArguments(
    output_dir="./results",                 # where checkpoints / logs are written

    # Batch sizes --------------------------------------------------------
    per_device_train_batch_size=16,          # micro-batch size for each GPU
    per_device_eval_batch_size=16,           # eval micro-batch size

    # Evaluation / saving cadence ---------------------------------------
    eval_steps=300,                          # run evaluation every N steps
    eval_strategy="steps",                 # trigger eval by step interval
    save_strategy="epoch",                 # save checkpoint at the end of each epoch
    num_train_epochs=1,                      # one full pass over the dataset

    # Optimiser & schedule ----------------------------------------------
    learning_rate=3e-5,
    lr_scheduler_type="cosine",            # cosine LR decay
    weight_decay=0.01,                       # AdamW weight-decay
    warmup_steps=50,                         # linear warm-up before decay kicks in

    # Mixed precision ----------------------------------------------------
    bf16=is_bfloat16_supported(),            # use bfloat16 if the GPU supports it
    fp16=not is_bfloat16_supported(),        # otherwise fall back to fp16

    # Misc ---------------------------------------------------------------
    gradient_accumulation_steps=1,           # no gradient accumulation here
    optim="adamw_torch",                   # PyTorch AdamW optimiser
    logging_steps=100,                       # log training loss every 100 steps
    report_to="wandb",                     # send metrics to Weights & Biases
)

# Build an SFTTrainer (supervised fine-tuning)
trainer = SFTTrainer(
    model=model,                             # the base LLM to be fine-tuned
    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["validation"],
    args=training_args,                      # hyper-parameters defined above
    data_collator=data_collator,             # handles padding / shifting
    tokenizer=tokenizer,                     # ensures correct special tokens
    #processing_class = tokenizer in the new versions
)

# Kick off the training loop
trainer.train()
```

**Normalize Numeric Answers**

```python
# --------------------------------------------------
# Normalise numeric answers inside the dataset
# --------------------------------------------------

def normalize_answer(x: str) -> str:
    """Return a *canonical* string representation of a number.

    Examples
    --------
    >>> normalize_answer("15.0")   # trailing .0 removed
    '15'
    >>> normalize_answer("3.50")   # trailing zero trimmed
    '3.5'
    >>> normalize_answer("-7.00")  # both decimal zeros trimmed
    '-7'
    Non‑numeric strings are returned unchanged (after .strip()).
    """
    try:
        f = float(x)  # attempt a numeric parse
    except ValueError:  # not a number → leave untouched
        return x.strip()

    # If the float is an *integer* value, emit without decimal part
    if f.is_integer():
        return str(int(f))
    # Otherwise, drop redundant trailing zeros / dot
    return str(f).rstrip("0").rstrip(".")

# Apply normalisation to every sample in a (HF) dataset
# -----------------------------------------------------
merged_dataset = merged_dataset.map(
    lambda ex: {"answer": normalize_answer(ex["answer"]) }
)
```

**Reward Functions**

```python
import re
from typing import List, Dict, Any, Optional
# --------------------------------------------------
# 1) Regex to capture *the last* boxed or "The Answer = …" number
#    that appears *after* the </think> tag in the assistant response.
# --------------------------------------------------
_NUM_RE = re.compile(
    r"</think>[\s\S]*?"                                   # skip everything up to the answer section
    r"(?:The\s+Answer\s*=\s*|\\boxed\s*[{(])"             # common maths-answer markers
    r"[^0-9+\-]*"                                         # ignore $, \\, spaces, punctuation
    r"([-+]?\d[\d\\!,]*(?:\.\d[\d\\!,]*)?)",              # capture the number itself
    flags=re.I | re.S,
)

# --------------------------------------------------
# 2) Primary reward – exact numeric match
# --------------------------------------------------

def answer_reward(completions, answer, **kwargs):
    """Return 1.0 if *numeric* answer matches the gold solution, else 0.0.
    `completions` / `answer` are lists-of-lists produced by TRL.
    """

    def _clean(raw):
        """Normalise a raw numeric string (strip $, \\ , ! , spaces, fix separators)."""
        if not raw:
            return None
        # Remove LaTeX, thousands separators, whitespace, currency symbols …
        raw = re.sub(r"[\\$!,\s]", "", raw)

        # Fix European vs US decimal / thousand separators
        if "," in raw and "." not in raw:
            raw = raw.replace(",", "") if re.fullmatch(r"-?\d{1,3}(,\d{3})+", raw) else raw.replace(",", ".")
        if "." in raw and re.fullmatch(r"-?\d{1,3}(\.\d{3})+", raw):
            raw = raw.replace(".", "")

        # Grab the *first* numeric chunk
        m = re.search(r"-?\d+(?:\.\d+)?", raw)
        if not m:
            return None
        num = float(m.group())
        # Remove trailing .0 (e.g. "2.0" → "2") and trailing zeros
        return str(int(num)) if num.is_integer() else str(num).rstrip("0").rstrip(".")

    rewards = []
    for comp, gold in zip(completions, answer):
        gold_num = _clean(str(gold))                             # canonical gold answer
        preds    = _NUM_RE.findall(comp[0]["content"])          # ALL numbers in the assistant answer
        pred_num = _clean(preds[-1]) if preds else None          # keep the *last* one
        rewards.append(1.0 if gold_num and pred_num and gold_num == pred_num else 0.0)
    return rewards

# --------------------------------------------------
# 3) Format reward – does the assistant follow <think> … </think> template?
# --------------------------------------------------

def format_reward(completions, **kwargs):
    pattern = re.compile(r"^<think>([\s\S]+?)</think>([\s\S]+)")
    responses = [c[0]["content"] for c in completions]
    matches   = [re.match(pattern, resp) for resp in responses]
    return [0.1 if m else 0.0 for m in matches]

# --------------------------------------------------
# 4) Number-coverage reward – does CoT mention the same numbers as question?
# --------------------------------------------------

def number_coverage_reward(prompts, completions, **kwargs):
    """Reward up to +0.1 for covering question numbers inside the <think> block."""
    rewards = []
    for p, comp in zip(prompts, completions):
        question_txt = p[1]["content"]                     # user message
        nums_q = set(re.findall(r"\d+", question_txt))    # all numbers in question

        think_block = re.search(r"<think>(.*?)</think>", comp[0]["content"], re.S)
        nums_think  = set(re.findall(r"\d+", think_block.group(1))) if think_block else set()

        recall = len(nums_q & nums_think) / (len(nums_q) or 1)  # fraction of question-numbers reproduced
        rewards.append(recall / 10)                             # scale: max ~0.1
    return rewards

# --------------------------------------------------
# 5) Reasoning length reward – incentivise *moderate* CoT length
# --------------------------------------------------

def reasoning_length_reward(completions, **kwargs):
    pattern   = re.compile(r"<think>([\s\S]+?)</think>")      # capture the CoT only
    tokenizer = kwargs["tokenizer"]                               # passed in from trainer
    rewards   = []

    for resp in [c[0]["content"] for c in completions]:
        m = pattern.search(resp)
        if not m:
            rewards.append(0.0)
            continue
        n_tok = len(tokenizer(m.group(1), add_special_tokens=False).input_ids)
        rewards.append(0.2 if 50 <= n_tok <= 400 else -0.1)      # short/long → penalty
    return rewards

# --------------------------------------------------
# 6) Explanation length reward – final answer size
# --------------------------------------------------

def explanation_length_reward(completions, **kwargs):
    pattern   = re.compile(r"</think>([\s\S]+)")              # text *after* </think>
    tokenizer = kwargs["tokenizer"]
    rewards   = []

    for resp in [c[0]["content"] for c in completions]:
        m = pattern.search(resp)
        if not m:
            rewards.append(0.0)
            continue
        n_tok = len(tokenizer(m.group(1), add_special_tokens=False).input_ids)
        rewards.append(0.1 if 40 <= n_tok <= 200 else -0.1)
    return rewards
```

**Eval Setup For Our GRPO Tuned Model**

```python
# --------------------------------------------------
# 0) Imports & global constants
# --------------------------------------------------
import re, random, pandas as pd, torch
from tqdm.auto import tqdm
from transformers import pipeline

# Generation hyper-parameters
MAX_NEW_TOKENS      = 1024
TEMPERATURE         = 0.5
TOP_P               = 0.9          # nucleus-sampling cutoff
TOP_K               = 20           # top-k cut-off
REPETITION_PENALTY  = 1.05
MAX_TOTAL_REWARD    = 1.5          # 1 + 0.1 + 0.1 + 0.2 + 0.1

# --------------------------------------------------
# 1) Shared helpers – number extraction / cleaning
# --------------------------------------------------
# Regex that finds the *last* boxed / “The Answer = …” number *after*
# the </think> tag (used by our reasoning format).
_NUM_RE = re.compile(
    r"</think>[\s\S]*?"
    r"(?:The\s+Answer\s*=\s*|\\boxed\s*[{(])"
    r"[^0-9+\-]*"
    r"([-+]?\d[\d\\!,]*(?:\.\d[\d\\!,]*)?)",
    flags=re.I | re.S,
)

def _clean(x: str | None) -> str | None:
    """
    Canonicalise a numeric string so that “$1,234.00” / “1 234,0” / etc.
    all map to the same clean representation.

    Steps
    -----
    1. Strip currency, back-slashes, thousands separators, spaces.
    2. Fix ',' / '.' ambiguities (locale-style separators).
    3. Return the *first* recognised float in the cleaned string.
    """
    if x is None:
        return None
    s = re.sub(r"[\\$!,\s]", "", str(x))

    # ---- Patch locale-style separators -----------------------------------
    if "," in s and "." not in s:  # e.g. "1,234" or "1,23"
        s = s.replace(",", "") if re.fullmatch(r"-?\d{1,3}(,\d{3})+", s) \
            else s.replace(",", ".")
    if "." in s and re.fullmatch(r"-?\d{1,3}(\.\d{3})+", s):  # e.g. "1.234.000"
        s = s.replace(".", "")
    # ----------------------------------------------------------------------

    m = re.search(r"-?\d+(?:\.\d+)?", s)
    if not m:
        return None
    f = float(m.group())
    return str(int(f)) if f.is_integer() else str(f).rstrip("0").rstrip(".")

def _extract_last_number(text: str) -> str | None:
    """Return the cleaned *last* numeric answer found in `text`."""
    hit = _NUM_RE.findall(text)
    return _clean(hit[-1]) if hit else None

# --------------------------------------------------
# 2) Reward functions
# --------------------------------------------------
def answer_reward(completions, answer, **_):
    """Primary reward – 1.0 if predicted number matches the gold, else 0.0."""
    scores = []
    for comp, gold in zip(completions, answer):
        gold_num = _clean(gold)
        pred_num = _extract_last_number(comp[0]["content"])
        scores.append(1.0 if gold_num and pred_num and gold_num == pred_num else 0.0)
    return scores

# → The rest are *auxiliary* shaping rewards used during RL
FMT_RE  = re.compile(r"^<think>[\s\S]+?</think>[\s\S]+", re.S)
def format_reward(completions, **_):
    """+0.1 if the assistant follows `<think> ... </think>` reasoning format."""
    return [0.1 if FMT_RE.match(c[0]["content"]) else 0.0 for c in completions]

def number_coverage_reward(prompts, completions, **_):
    """
    + up to 0.1 if the chain-of-thought mentions most of the numbers that
    appear in the question (encourages copying key figures).  We normalise by 10.
    """
    out = []
    for p, c in zip(prompts, completions):
        q = set(re.findall(r"\d+", p[1]["content"]))       # numbers in question
        t = re.search(r"<think>(.*?)</think>", c[0]["content"], re.S)
        t = set(re.findall(r"\d+", t.group(1))) if t else set()
        out.append(len(q & t) / 10 / (len(q) or 1))
    return out

def _ensure_tok(kwargs):
    """Tiny helper – pull tokenizer from kwargs or fall back to global."""
    return kwargs.get("tokenizer", globals()["tokenizer"])

def reasoning_length_reward(completions, **kwargs):
    """
    +0.2 if chain-of-thought is between 50 and 450 tokens long
    (very short 🤔   → not enough reasoning,
     very long  📚→ waffle / waste of tokens).
    """
    tok = _ensure_tok(kwargs); out = []
    for c in completions:
        m = re.search(r"<think>(.*?)</think>", c[0]["content"], re.S)
        n = len(tok(m.group(1), add_special_tokens=False).input_ids) if m else 0
        out.append(0.2 if 50 <= n <= 450 else 0.0)
    return out

def explanation_length_reward(completions, **kwargs):
    """
    +0.1 if the *final answer section* (after </think>) is 50-200 tokens.
    Ensures we don’t put the entire reasoning in the public answer.
    """
    tok = _ensure_tok(kwargs); out = []
    for c in completions:
        after = c[0]["content"].split("</think>", 1)[-1] if "</think>" in c[0]["content"] else ""
        n = len(tok(after, add_special_tokens=False).input_ids)
        out.append(0.1 if 50 <= n <= 200 else 0.0)
    return out

# --------------------------------------------------
# 3) Generation helper – wraps the HF pipeline
# --------------------------------------------------
gen_pipe = pipeline(
    "text-generation",
    model=final_model,
    tokenizer=tokenizer,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)

def run_model(messages):
    """One-shot helper to call the model with the chosen sampling settings."""
    out = gen_pipe(
        messages,
        max_new_tokens   = MAX_NEW_TOKENS,
        temperature      = TEMPERATURE,
        top_p            = TOP_P,
        top_k            = TOP_K,
        do_sample        = True,
        repetition_penalty = REPETITION_PENALTY,
    )
    return out[0]["generated_text"][-1]["content"]

# --------------------------------------------------
# 4) Evaluate *one* example
# --------------------------------------------------
def evaluate_example(ex):
    assistant = run_model(ex["prompt"])
    comp  = [[{"role": "assistant", "content": assistant}]]
    prom  = [ex["prompt"]]
    truth = [ex["answer"]]

    # --- compute individual reward components
    ans_r  = answer_reward(comp, truth)[0]
    fmt_r  = format_reward(comp)[0]
    num_r  = number_coverage_reward(prom, comp)[0]
    reas_r = reasoning_length_reward(comp, tokenizer=tokenizer)[0]
    expl_r = explanation_length_reward(comp, tokenizer=tokenizer)[0]

    total  = ans_r + fmt_r + num_r + reas_r + expl_r
    return {
        "question"             : ex["question"],
        "ground_truth"         : ex["answer"],
        "model_answer"         : assistant,
        "model_numeric_answer" : _extract_last_number(assistant),
        "answer_reward"        : ans_r,
        "format_reward"        : fmt_r,
        "numcov_reward"        : num_r,
        "reason_len_reward"    : reas_r,
        "expl_len_reward"      : expl_r,
        "total_reward"         : total,
        "relative_reward"      : total / MAX_TOTAL_REWARD,
    }

# --------------------------------------------------
# 5) Bulk evaluation on a random subset
# --------------------------------------------------
def score_examples(dataset_split, num_examples=300, seed=42):
    random.seed(seed)
    idxs = random.sample(range(len(dataset_split)), num_examples)
    rows = [evaluate_example(dataset_split[i]) for i in tqdm(idxs, desc="Scoring")]
    return pd.DataFrame(rows)

# --------------------------------------------------
# 6) Run evaluation on GSM-8K test split
# --------------------------------------------------
final_model_scores = score_examples(dataset_gsm8k["test"])
```

**Eval Setup for Qwen2.5-7B Inst Model**

```python
# --------------------------------------------------
# Imports & global settings
# --------------------------------------------------
import re
import random
import pandas as pd
import torch
from tqdm.auto import tqdm
from transformers import pipeline

# Generation hyper-parameters
MAX_NEW_TOKENS     = 1024  # maximum length the model may append
TEMPERATURE        = 0.5   # sampling temperature (higher = more randomness)
TOP_P              = 0.9   # nucleus sampling threshold
TOP_K              = 20    # top-k sampling threshold
REPETITION_PENALTY = 1.05  # discourage the model from repeating itself

# --------------------------------------------------
# Regex helper  (no explicit THINK tag!)
# --------------------------------------------------
NUM_PATTERN = re.compile(
    r"(?:The\s+Answer\s*=\s*|\\boxed\s*[{(])"  # common answer markers
    r"[^0-9+\-]*"                                   # skip $, \\, spaces, etc.
    r"([-+]?\d[\d\\!,]*(?:\.\d[\d\\!,]*)?)", # capture the number itself
    flags=re.I | re.S,
)

# --------------------------------------------------
# Utility functions
# --------------------------------------------------

def _clean_number(raw: str | None) -> str | None:
    """Strip LaTeX / formatting noise and normalise the numeric string."""
    if raw is None:
        return None
    # remove LaTeX, thousand-separators, spaces
    s = re.sub(r"[\\$!,\s]", "", raw)
    # fix comma / dot issues (european vs US notation)
    if "," in s and "." not in s:
        # all commas are thousand-separators
        s = s.replace(",", "") if re.fullmatch(r"-?\d{1,3}(,\d{3})+", s) else s.replace(",", ".")
    if "." in s and re.fullmatch(r"-?\d{1,3}(\.\d{3})+", s):
        s = s.replace(".", "")
    # grab first numeric chunk
    m = re.search(r"-?\d+(?:\.\d+)?", s)
    if not m:
        return None
    f = float(m.group())
    return str(int(f)) if f.is_integer() else str(f).rstrip("0").rstrip(".")


def _extract_last_number(text: str) -> str | None:
    """Return the last numeric answer the model produced (already cleaned)."""
    hits = NUM_PATTERN.findall(text)
    return _clean_number(hits[-1]) if hits else None

# --------------------------------------------------
# Generation helper
# --------------------------------------------------

gen_pipe = pipeline(
    "text-generation",
    model=base_inst_model,      # Qwen2.5-7B Inst model (Before fine tuning)
    tokenizer=tokenizer,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)


def run_model(messages):
    """Call the text-generation pipeline with the chosen sampling setup."""
    out = gen_pipe(
        messages,
        max_new_tokens=MAX_NEW_TOKENS,
        temperature=TEMPERATURE,
        top_p=TOP_P,
        top_k=TOP_K,
        do_sample=True,
        repetition_penalty=REPETITION_PENALTY,
    )
    return out[0]["generated_text"][-1]["content"]

# --------------------------------------------------
# Single-sample evaluation
# --------------------------------------------------

def evaluate_example(ex):
    """Run the model on one example and collect numeric answer."""
    reply = run_model(ex["prompt"])
    return {
        "question"            : ex["question"],
        "ground_truth"        : ex["answer"],
        "model_answer"        : reply,
        "model_numeric_answer": _extract_last_number(reply),
    }

# --------------------------------------------------
# Batch evaluation utilities
# --------------------------------------------------

def score_examples(dataset_split, num_examples: int = 300, seed: int = 42):
    """Pick *num_examples* random questions, evaluate, return a DataFrame."""
    random.seed(seed)
    idxs = random.sample(range(len(dataset_split)), num_examples)
    rows = [evaluate_example(dataset_split[i]) for i in tqdm(idxs, desc="Scoring")]
    return pd.DataFrame(rows)

# --------------------------------------------------
# Run the evaluation on GSM8K test split using the base model
# --------------------------------------------------

df_scores_base_model = score_examples(dataset_gsm8k["test"])
```
